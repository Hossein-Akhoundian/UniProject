### Load Dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../Dataset/results.csv")

df.head()
```

```python
# حذف داده های ناقص
df = df.dropna()
```

```python
# نبدیل تاریخ
df['date'] = pd.to_datetime(df['date'])
```

```python
df.info()
```

```python
# missing values (check)
df.isnull().sum()
```

```python
df['Year'] = df['date'].dt.year
```

```python
df['Month'] = df['date'].dt.month
```

```python
df['Year'].value_counts().sort_index().plot(figsize=(12,4))
```
